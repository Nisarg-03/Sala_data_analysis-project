# Sales Data Analysis Project
This project analyzes retail sales data to understand customer behavior, product performance, and sales trends.

```python
# import Libraries
import pandas as pd 
import matplotlib.pyplot as plt 

```

```python
# Load dataset
data = pd.read_csv("/Users/nisarg/Downloads/Sales_data_analysis/sample_sales_data.csv")
print(data.head())
```

```python
# Data Exploration

print(data.shape)
print(data.columns)
print(data.info())
print(data.describe())



```

```python
# Data Cleaning

print(data.isnull().sum())
print(data.dropna())
```

```python
# Calculate Total Sales
Total_sales = data["Total"].sum()

print("Total_sales:",Total_sales)
```

```python
# Sales by Product line
Product_sales = data.groupby("Product line")["Total"].sum()

print(Product_sales)
```

```python
# Sales by City
City_sales = data.groupby("City")["Total"].sum()

print(City_sales)
```

```python
# Best Payment Method 
payment_sales = data.groupby("Payment")["Total"].sum()

print(payment_sales)
```

```python
# bar chart of Sales by Product Line
Product_sales.plot(kind="bar")

plt.title("Sales by Product line")
plt.xlabel("Product line")
plt.ylabel("revenue")

plt.show()
```

```python
# Bar chart of Sales by City
City_sales.plot(kind="bar")

plt.title("Sales by City")
plt.show()
```

```python
# Best Product Category
Top_products = Product_sales.sort_values(ascending=False)

print(Top_products)
```

```python
# Average Rating by Product line
Rating_analysis = data.groupby("Product line")["Rating"].mean()

print(Rating_analysis)
```

```python
data["Date"] = pd.to_datetime(data["Date"])

Sales_trend = data.groupby("Date")["Total"].sum()

Sales_trend.plot(kind="line")
plt.title("Sales Trend over Time")
plt.show()
```

```python
# Anlyze sales by Branch
branch_sales = data.groupby("Branch")["Total"].sum()

print(branch_sales)

```

```python
branch_sales.plot(kind="bar")

plt.title("Sales by Branch")
plt.xlabel("Branch")
plt.ylabel("Revenue")

plt.show()
```
